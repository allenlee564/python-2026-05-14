````python
**程式流程：**

1. 輸入身高（公分）與體重（公斤）
2. 計算 BMI = 體重 ÷ (身高 ÷ 100)²
3. 依照 BMI 值判斷健康狀態並給予建議
4. 輸出格式化報告

**BMI 分類：**| BMI 範圍 | 狀態 | 建議 |
| --- | --- | --- |
| < 18.5 | 體重過輕 | 請適量增加營養攝取 |
| 18.5 ～ 24.9 | 正常範圍 | 保持良好生活習慣 |
| 25.0 ～ 29.9 | 體重過重 | 建議適度運動與飲食控制 |
| ≥ 30.0 | 肥胖 | 請諮詢醫師進行健康管理 |

- BMI 值保留兩位小數
- 欄位標籤對齊（使用全形空格或 f-string 對齊）
- 分隔線固定 30 個字元寬
````

````python
print("======== BMI 健康分析 ========")
input_str = input("請輸入身高（公分）：")
height_cm = float(input_str)
input_str = input("請輸入體重（公斤）：")
weight_kg = float(input_str)
bmi = weight_kg / (height_cm / 100) ** 2
bmi_rounded = round(bmi, 2)
print(f"身高：{height_cm} cm")
print(f"體重：{weight_kg} kg")
print(f"BMI：{bmi_rounded}")
if bmi < 18.5:
    print("狀態：體重過輕")
    print("建議：請適量增加營養攝取")
elif 18.5 <= bmi < 25:
    print("狀態：正常範圍")
    print("建議：保持良好生活習慣")
elif 25 <= bmi < 30:
    print("狀態：體重過重")
    print("建議：建議適度運動與飲食控制")
else:
    print("狀態：肥胖")
    print("建議：請諮詢醫師進行健康管理")
print("==============================")
````

````python
電影院需要一套售票程式，依照觀眾年齡與會員身份計算票價。

**票價規則：**

| 類別 | 年齡條件 | 原價 |
| --- | --- | --- |
| 兒童票 | 未滿 6 歲 | 免費 |
| 優待票 | 6 歲以上，未滿 12 歲 | 150 元 |
| 學生票 | 12 歲以上，未滿 18 歲 | 220 元 |
| 全票 | 18 歲以上，未滿 65 歲 | 350 元 |
| 敬老票 | 65 歲以上 | 150 元 |

**附加規則：**

- 會員享有 85 折優惠（免費票不適用）
- 週末（使用者輸入 `y` 代表週末）加收 50 元假日附加費（免費票不適用）

**輸出格式範例：**

```
========== 售票明細 ==========
票種：全票
原價：350 元
會員折扣：-52 元（85折）
假日附加費：+50 元
應付金額：348 元
==============================
```
````

````python
is_weekend = input("是否為週末？(y/n)：").lower() == 'y'
age = int(input("請輸入年齡："))
match age:
    case age if age < 6:
        ticket_type = "兒童票"
        original_price = 0
    case age if 6 <= age < 12:
        ticket_type = "優待票"
        original_price = 150
    case age if 12 <= age < 18:
        ticket_type = "學生票"
        original_price = 220
    case age if 18 <= age < 65:
        ticket_type = "全票"
        original_price = 350
    case age if age >= 65:
        ticket_type = "敬老票"
        original_price = 150
is_member = input("是否為會員？(y/n)：").lower() == 'y'
discount = 0
if is_member and original_price > 0:
    discount = original_price * 0.15
holiday_fee = 50 if is_weekend and original_price > 0 else 0
final_price = original_price - discount + holiday_fee
print("========== 售票明細 ==========")
print(f"票種：{ticket_type}")
print(f"原價：{original_price} 元")
print(f"會員折扣：-{discount:.0f} 元（85折）")
print(f"假日附加費：+{holiday_fee} 元")
print(f"應付金額：{final_price:.0f} 元")
print("==============================")
````
