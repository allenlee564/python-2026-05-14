````python
印出 1 到 50 的數字，但：

- 3 的倍數改印 `Fizz`
- 5 的倍數改印 `Buzz`
- 同時為 3 與 5 的倍數改印 `FizzBuzz`
- 每個數字以空格分隔，同一行輸出
````

````python
for i in range(1, 51):
    if i% 15 == 0:
        print("FizzBuzz", end=" ")
    if i % 3 ==0:
        print("fizz", end=" ")
    elif i % 5 ==0:
        print("buzz", end=" ")
    else:
        print(i, end=" ")
````

````python
印出完整的九九乘法表，格式對齊如下：

```
1x1= 1  1x2= 2  1x3= 3  1x4= 4  1x5= 5  1x6= 6  1x7= 7  1x8= 8  1x9= 9
2x1= 2  2x2= 4  2x3= 6  ...
...
9x1= 9  9x2=18  ...  9x9=81
```

**要求：**

- 每個算式固定寬度（使用 f-string 對齊）
- 每列結束後換行
````

````python
for i in range(1, 10):
    for j in range(1, 10):
        print(f"{i}*{j}={i*j:2}", end=" ")
    print()  # 換行
    
````

````python
顧客在便利商店使用自助結帳機，系統需管理購物車內容。

**商品目錄（字典）：**

```python
catalog = {
    "御飯糰": 35,
    "礦泉水": 20,
    "布丁": 25,
    "關東煮": 15,
    "飯糰": 30,
}
```

**系統功能（使用 `while` 迴圈選單）：**

1. 新增商品：輸入商品名稱與數量，若商品不在目錄中提示錯誤；購物車以字典儲存 `{商品名稱: 數量}`
2. 移除商品：輸入商品名稱，從購物車移除（不存在時提示）
3. 查看購物車：列出所有商品、各項小計，並顯示總金額
4. 結帳：顯示明細，若購買 3 件以上享 95 折，輸出折扣資訊與最終金額
5. 離開：輸入 `0` 結束

**結帳輸出格式：**

```
====== 結帳明細 ======
御飯糰    x2   70 元
礦泉水    x1   20 元
----------------------
小計：90 元
折扣：-4 元（95折）
總計：86 元
======================
```
````

````python
catalog = {
    "御飯糰": 35,
    "礦泉水": 20,
    "布丁": 25,
    "關東煮": 15,
    "飯糰": 30,
}

cart = {}

while True:
    print("1. 新增商品")
    print("2. 移除商品")
    print("3. 查看購物車")
    print("4. 結帳")
    print("輸入0離開")
    choice = input("請選擇操作： ")
    if choice == "0":
        break
    if choice == "1":
        print("可選商品如下(輸入商品名稱要完全一致)：")
        print(catalog)
        pname = input("請輸入商品名稱：").strip()
        if pname not in catalog:
            print("商品不存在，請重新輸入")
        QTY = int(input("請輸入數量："))
        cart[pname] = cart.get(pname, 0) + QTY
         # cart[pname] = QTY  # 這樣會覆蓋原有數量
         # cart[pname] = cart.get(pname, 0) + QTY
    print("已放入購物車")

    if choice == "2":
        paname = input("請輸入要移除的商品名稱：").strip()
        if paname  in cart:
            del cart[paname]
            print("商品已從購物車中移除")

    if choice == "3":
        print("=" * 40)
        if not cart:
            print("購物車是空的")
        for name, qty in cart.items():
            print(f"{name} x{qty} {catalog[name] * qty} 元")

    if choice == "4":
        print("====== 結帳明細 ======")
        total = 0
        total_qty = 0
        for name, qty in cart.items():
            subtotal = catalog[name] * qty
            print(f"{name}    x{qty}   {subtotal} 元")
            total += subtotal
            total += qty

        print(f"小計:{total}")
        if total_qty >= 3:
            dc = total * 0.05
            print(f"折扣：{dc * -1}")
            print(f"總計：{total - dc}")
        else:
            print(f"總計：{total}")
        
````
